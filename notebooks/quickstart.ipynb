# Checkpoint Courier — CPU demo
Run both cells. This synthetic demonstration needs no GPU or Drive access. It uses temporary storage, not a persistent backup. For real training, mount Drive and follow the README; cloud durability is not tested by this demo.

```python
import subprocess, sys
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', 'git+https://github.com/B-jack7/checkpoint-courier.git@main'])
```

```python
from pathlib import Path
from tempfile import TemporaryDirectory
from checkpoint_courier import Store

with TemporaryDirectory() as directory:
    root = Path(directory)
    original = root / "epoch-15.pt"
    original.write_bytes(b"synthetic checkpoint bytes\n" * 1000)
    store = Store(root / "backups")
    saved = store.save(original, tag="epoch-15")
    original.unlink()
    restored = store.restore(saved["id"], root / "recovered.pt")
    assert restored.read_bytes() == b"synthetic checkpoint bytes\n" * 1000
    print("Recovery verified:", store.verify(saved["id"])["sha256"])
```
